# **Model Optimization**

Model optimization refers to a collection of techniques and processes used to improve a model's performance, efficiency, and suitability for deployment—especially on resource-constrained environments like mobile or edge devices.

It can refer to two broad areas:
1. **Training-Time Optimization (Improving learning and accuracy)**
  - This involves tuning the model during training to improve convergence, generalization, and performance.
2. **Inference-Time Optimization (Improving speed, size, and efficiency)**
  - This focuses on making trained models more efficient for deployment, particularly on edge or mobile devices.


# **Training-Time Optimization**

Training-time optimization in deep learning refers to the strategies used to help a model learn more effectively, faster, and generalize better. It focuses on improving the training process so that the final model performs well on unseen data. Here's a breakdown of the key components and techniques involved:
- Loss function minimization using optimizers like SGD, Adam, etc.
- Hyperparameter tuning (learning rate, batch size, etc.)
- Regularization (dropout, weight decay)
- Batch normalization
- Early stopping to avoid overfitting


## Loss function minimization
At the core of training is minimizing a loss function, which measures how far off the model’s predictions are from the true labels.

Common loss functions:
- Cross-entropy for classification
- Mean squared error (MSE) for regression

The goal of training is to adjust the model weights to minimize the loss over the training set.

## Optimizers
Optimizers determine how model weights are updated during training.

- SGD (Stochastic Gradient Descent): Basic optimizer using gradient direction.
- Momentum: Helps accelerate SGD by remembering past gradients.
- Adam: Popular adaptive optimizer that adjusts learning rates per parameter.
- Others: RMSProp, AdaGrad, etc.

Each optimizer has hyperparameters (like learning rate) that heavily influence performance.

## Learning Rate Scheduling

The learning rate controls how big the updates to model weights are.

- If it’s too high → training may diverge.
- If it’s too low → training is slow or stuck.

Schedulers adjust the learning rate over time, e.g.:

- Step decay: Reduce LR at set epochs
- Cosine annealing or exponential decay
- Warmup: Gradually increase LR at the start to stabilize training

## Regularization

**Used to prevent overfitting and improve generalization.**

- Dropout: Randomly disables neurons during training
- Weight decay (L2 regularization): Penalizes large weights
- Early stopping: Stop training when validation loss starts increasing
- Data augmentation: Artificially enlarges the dataset (especially useful for images)

## Batch Normalization & Layer Normalization
Helps stabilize and speed up training by normalizing activations across mini-batches (or within layers).

- BatchNorm is widely used in CNNs
- LayerNorm is often used in transformers or RNNs

Benefits:

- Reduces internal covariate shift
- Allows higher learning rates
- Acts as a regularizer

## Hyperparameter Tuning

Many settings affect training and performance, including:

- Learning rate
- Batch size
- Number of layers/neurons
- Optimizer type
- Activation functions

Tuning these (manually or using tools like Optuna, Ray Tune, or Grid/Random Search) can greatly improve training efficiency.

## Training Tricks and Best Practices

- Gradient clipping: Prevent exploding gradients
- Label smoothing: Softens the one-hot labels, helps generalization
- Mixup / CutMix: Advanced augmentation strategies
- Use validation set: Monitor generalization and avoid overfitting
- Weight initialization: Choose good initial weights (Xavier, He initialization)


| Technique                    | Purpose                     |
| ---------------------------- | --------------------------- |
| Optimizer (Adam, SGD, etc.)  | Guides weight updates       |
| Learning Rate Scheduler      | Adapts learning speed       |
| Regularization (Dropout, L2) | Prevents overfitting        |
| Batch/Layer Norm             | Stabilizes training         |
| Early stopping               | Avoids overfitting          |
| Data augmentation            | Improves generalization     |
| Hyperparameter tuning        | Maximizes model performance |


# **Inference-Time Optimization**

Inference-time optimization in deep learning focuses on making trained models faster, smaller, and more efficient during deployment—especially on edge devices like smartphones, embedded systems, or IoT devices where compute power and memory are limited.

Key techniques:

- Quantization: Reducing precision (e.g., from float32 to int8) to lower memory and compute cost.
- Pruning: Removing unnecessary weights or neurons to make the model smaller and faster.
- Knowledge Distillation: Training a smaller "student" model to mimic a larger "teacher" model.
- Operator fusion / graph optimization: Combining operations to reduce memory access and latency.
- Threading/parallelism and hardware acceleration: Leveraging multi-core CPUs, GPUs, or NPUs efficiently.


## Quantization

Reduce the precision of model weights and activations (e.g., from float32 to int8 or float16).

Types:

- Post-training quantization (PTQ): Applied after training.
- Quantization-aware training (QAT): Simulates quantization during training for better accuracy.

Benefits:

- 4x smaller models (float32 → int8)
- 2–4x faster inference
- Lower power usage

## Pruning
**Remove unnecessary weights or neurons** (e.g., small-valued weights that contribute little).

Types:

- Unstructured pruning: Prunes individual weights (sparser networks).
- Structured pruning: Removes full channels/filters (hardware-friendly).

Benefit:

Smaller, faster models with minimal accuracy loss.

## Knowledge Distillation

Train a small “student” model to mimic a large “teacher” model.

Benefit:

- Smaller model that preserves much of the accuracy of the large one.
Process:

- Train a small model using a combination of:
  - Ground-truth labels
  - Soft outputs from the teacher model

## Operator Fusion / Graph Optimization

Merge operations into a single optimized kernel (e.g., fuse Conv2D + BatchNorm + ReLU).

Benefits:

- Reduces memory access
- Increases throughput on CPUs/GPUs


Tools:

- TorchScript
- ONNX Runtime
- TensorRT
- TVM
- TFLite Converter


## Model Compilation / Conversion

Convert and optimize the model to run efficiently on the target hardware.

| Framework       | Target                                      |
| --------------- | ------------------------------------------- |
| **TorchScript** | Optimized PyTorch model for C++/mobile      |
| **ONNX**        | Cross-framework inference                   |
| **TFLite**      | TensorFlow Lite (Android, microcontrollers) |
| **TensorRT**    | NVIDIA GPU acceleration                     |


## Parallelism & Threading

Leverage multiple CPU cores for inference.

## Model Architecture Design for Efficiency

Use models designed for speed/efficiency from the start:

- MobileNetV2/V3
- EfficientNet-Lite
- Tiny YOLO / YOLO-NAS-S
- DistilBERT / TinyBERT for NLP


## Summary

| Optimization Technique            | Goal                            | When to Use                |
| --------------------------------- | ------------------------------- | -------------------------- |
| Quantization                      | Reduce model size & latency     | Best for edge/mobile       |
| Pruning                           | Remove redundant weights        | After training             |
| Knowledge Distillation            | Compress model w/ accuracy      | During/after training      |
| Graph Optimization                | Fuse layers, reduce ops         | During conversion/export   |
| Compilation (TorchScript, TFLite) | Hardware-specific optimizations | Before deployment          |
| Threading/Parallelism             | Speed up CPU inference          | At runtime (e.g., Android) |
| Efficient Architectures           | Design for low-cost inference   | From the start             |


## **Tools for Inference Optimization**

| Tool            | Purpose                        | Platform                     |
| --------------- | ------------------------------ | ---------------------------- |
| **TorchScript** | PyTorch model export & fusion  | PyTorch mobile               |
| **ONNX**        | Cross-platform model exchange  | CPU/GPU/Edge                 |
| **TFLite**      | Optimized TF models for mobile | Android/iOS/microcontrollers |
| **TensorRT**    | NVIDIA GPU acceleration        | Server/GPU edge              |

# **Typical Optimization Pipeline for Edge Deployment**

- Train a full-precision model
- Apply quantization/pruning/distillation
- Convert/export using TFLite / TorchScript / ONNX
- Fuse operators if supported
- Tune threading, batching for target hardware
- Deploy to edge device

# **Gradle**

Gradle is the build automation system used in Android development. It manages how your app is compiled, built, and packaged. In Android Studio, Gradle is the tool that orchestrates the entire build process — from code compilation to generating the final .apk or .aab (Android App Bundle).

## What Gradle Does in Android:

| Task                      | Role                                                      |
| ------------------------- | --------------------------------------------------------- |
| **Compiling code**        | Translates Java/Kotlin/XML into `.dex` files              |
| **Dependency management** | Downloads and manages libraries (e.g., PyTorch, Retrofit) |
| **Build variants**        | Supports different flavors like debug/release             |
| **Packaging**             | Assembles resources, assets, code into an APK/AAB         |
| **Signing**               | Signs your app for release or debug                       |
| **Testing**               | Runs unit tests and instrumentation tests                 |


## Key Gradle Files in Android Projects

1. ```settings.gradle ```
- Defines which modules are included in your build.

2. ```build.gradle (Project level)```
- Contains settings for the whole project, like the Gradle plugin version and repositories.

3. ```build.gradle (App level)```
- Defines SDK versions, dependencies, version code/name, and build types.

Example:

```gradle
android {
    compileSdk 33
    defaultConfig {
        applicationId "com.example.myapp"
        minSdk 21
        targetSdk 33
        versionCode 1
        versionName "1.0"
    }
    buildTypes {
        release {
            minifyEnabled true
            proguardFiles getDefaultProguardFile('proguard-android-optimize.txt'), 'proguard-rules.pro'
        }
    }
}

dependencies {
    implementation 'org.pytorch:pytorch_android_lite:1.13.0'
}

```

## 2. MainActivity.java

Basic image classification using a Bitmap

```java
package com.example.mypytorchapp;

import android.app.Activity;
import android.graphics.Bitmap;
import android.graphics.BitmapFactory;
import android.os.Bundle;
import android.widget.TextView;

import org.pytorch.IValue;
import org.pytorch.Module;
import org.pytorch.Tensor;
import org.pytorch.torchvision.TensorImageUtils;

import java.io.File;
import java.io.FileOutputStream;
import java.io.InputStream;

public class MainActivity extends Activity {

    private Module module;

    @Override
    protected void onCreate(Bundle savedInstanceState) {
        super.onCreate(savedInstanceState);
        setContentView(R.layout.activity_main);

        TextView resultText = findViewById(R.id.resultText);

        try {
            module = Module.load(assetFilePath("model.pt"));

            Bitmap bitmap = BitmapFactory.decodeStream(getAssets().open("sample.jpg"));
            Tensor inputTensor = TensorImageUtils.bitmapToFloat32Tensor(
                    bitmap,
                    TensorImageUtils.TORCHVISION_NORM_MEAN_RGB,
                    TensorImageUtils.TORCHVISION_NORM_STD_RGB
            );

            Tensor outputTensor = module.forward(IValue.from(inputTensor)).toTensor();
            float[] scores = outputTensor.getDataAsFloatArray();

            int maxIndex = 0;
            for (int i = 1; i < scores.length; i++) {
                if (scores[i] > scores[maxIndex]) maxIndex = i;
            }

            resultText.setText("Predicted class: " + maxIndex);

        } catch (Exception e) {
            resultText.setText("Error: " + e.getMessage());
        }
    }

    private String assetFilePath(String assetName) throws Exception {
        File file = new File(getFilesDir(), assetName);
        if (file.exists() && file.length() > 0) return file.getAbsolutePath();

        try (InputStream is = getAssets().open(assetName);
             FileOutputStream os = new FileOutputStream(file)) {
            byte[] buffer = new byte[4 * 1024];
            int read;
            while ((read = is.read(buffer)) != -1) os.write(buffer, 0, read);
            os.flush();
        }

        return file.getAbsolutePath();
    }
}

```
