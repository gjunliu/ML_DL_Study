# What is ResNet
ResNet (Residual Network) is a deep convolutional neural network architecture introduced by Microsoft Research in 2015. It became famous after winning the ImageNet 2015 competition by a large margin. Its key innovation is the use of "residual connections", also called skip connections, which make it easier to train very deep networks.

## The Core Idea of ResNet
Instead of learning a direct mapping from input to output:

Desired output:     $H(x)$

Traditional NN:     $H(x) = F(x)$

ResNet learns a residual function:

ResNet:             $H(x) = F(x) + x$

where:
- $F(x)$ is the output of a few convolutional layers
- $x$ is the original input to those layers
- The output is $F(x) + x$

```
        x
        │
   ┌────┴────┐
   │         │
[Conv1]    (Skip)
   ↓         │
[BN1]        │
   ↓         │
[ReLU]       │
   ↓         │
[Conv2]      │
   ↓         │
[BN2]        │
   └────┬────┘
        │
     [Addition]
        ↓
     [ReLU]
        ↓
      Output

```


## Why Use Residual Connections?
Deep networks often suffer from:
- **Vanishing gradients** (hard to train deep layers)
- **Degradation** (accuracy gets worse as depth increases)

Residual connections help by:
- Allowing gradients to flow directly through skip paths during backpropagation
- Letting the model "skip" layers if needed — so it’s easier to learn identity mappings



# Residual Block

A **residual block** looks like this:
```
Input (x)
   │
[Conv → ReLU → Conv]
   │
Add shortcut (x)
   │
Output (y = F(x) + x)

```

Where:
- $F(x)$ is the learned transfomation (e.g., two Conv layers)
- $x$ is the orginal input
- The final output is $F(x) + x$ (called the residual connection)

This makes the network learn only the **difference (residual)** from the identity.



# Batch Normalization

**Batch Normalization** is a technique to normalize the inputs of each layer, so that they have a mean ≈ 0 and variance ≈ 1, computed per mini-batch.

- Normalizaiton: Collapse inputs to be between 0 and 1
- Standardization: make mean 0 and variance 1

Let us use $B$ to denote a mini-batch of size $m$ of the entire training set. The empirical mean and variance of B could thus be denoted as:

$$
\mu_B = \frac{1}{m}\sum\limits_{i = 1}^{m}x_i \\
σ^2_{B} = \frac{1}{m}∑\limits_{i = 1}^{m}(x_i - \mu_B)^2
$$

For a layer of the network with d-dimensional input, $x = (x^{(1)}, …, x^{(d)})$, each dimension of its input is then normalized (i.e. re-centered and re-scaled) separately,

$$
\hat x_i^{(k)} = \frac{x_i^{(k)} - \mu_B^{(k)}}{\sqrt{(\mu_B^{(k)})^2} + ϵ}
$$

where $k \in [1, d]$ and $i \in [1, m]$; $\mu_B^{(k)}$ and $σ_B^{(k)}$ are the per-dimension mean and standard deviation, respectively.
$ϵ$ is added in the denominator for numerial stability and is an arbitarily small constant.
